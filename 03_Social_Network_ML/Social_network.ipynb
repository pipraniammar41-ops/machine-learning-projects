```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings 
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("/kaggle/input/datasets/rakeshrau/social-network-ads/Social_Network_Ads.csv")
```

```python
df.drop(["User ID"],axis=1)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["Gender"] = le.fit_transform(df["Gender"])
```

```python
df["Gender"]
```

```python
df.head()
```

```python
X= df.drop("Purchased", axis=1)
y=df["Purchased"]
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y, test_size= 0.20, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test,y_pred)
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(y_test, y_pred)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
X2= df.drop(["Purchased","Gender"],axis=1)
y = df["Purchased"]
```

```python
from sklearn.model_selection import train_test_split

X2_train,X2_test,y_train,y_test=train_test_split(X2,y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression

model=LogisticRegression()
model.fit(X2_train,y_train)
```

```python
y_pred_2 = model.predict(X2_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_pred_2,y_test)
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(y_test, y_pred_2)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_2))
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

```python
X2_train_scaled = scaler.fit_transform(X2_train)
X2_test_scaled = scaler.transform(X2_test)
```

```python
from sklearn.linear_model import LogisticRegression

model_1= LogisticRegression()
model_1.fit(X2_train_scaled,y_train)
```

```python
y_pred_3 = model_1.predict(X2_test_scaled)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_pred_3,y_test)
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(y_test,y_pred_3)

```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_3))
```
